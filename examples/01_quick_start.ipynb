## Some data, in plain Python

```python
import numpy as np
import pandas as pd
import econenv

rng = np.random.default_rng(20260904)
n = 120
df = pd.DataFrame({'x1': rng.normal(size=n), 'x2': rng.normal(size=n)})
df['y'] = 2.0 + 0.5 * df.x1 - 0.3 * df.x2 + rng.normal(scale=0.4, size=n)
df.head()
```

## R

`-i df` pushes the frame in first.

```python
%%R -i df
fit <- lm(y ~ x1 + x2, data = df)
summary(fit)
```

## Stata

```python
%stata_push df
```

```python
%%stata
regress y x1 x2
```

## EViews

```python
%%eviews -i df
equation eq1.ls y c x1 x2
```

```python
%eviews_get eq1.@r2
```
