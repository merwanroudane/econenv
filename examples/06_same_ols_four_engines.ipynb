# 6 - The same OLS in four engines

The reason the project exists (brief section 48). Run one specification
everywhere and get a single comparison table - with the differences
**explained**, not smoothed over.

```python
%load_ext econenv
import econenv
```

```python
import numpy as np
import pandas as pd
import econenv

rng = np.random.default_rng(20260904)
n = 120
df = pd.DataFrame({'x1': rng.normal(size=n), 'x2': rng.normal(size=n)})
df['y'] = 2.0 + 0.5 * df.x1 - 0.3 * df.x2 + rng.normal(scale=0.4, size=n)
df.head()
```

## One call

```python
comparison = econenv.compare_ols(df, 'y ~ x1 + x2')
comparison
```

## The pieces

```python
comparison.coefficients()
```

```python
comparison.summary()
```

## Do they agree?

Both an absolute and a relative tolerance are applied - a 1e-15 gap is
floating-point arithmetic, not a finding.

```python
comparison.differences()
```

```python
print('agree:', comparison.agree)
```

## Why the information criteria differ

They are different normalisations of the same quantity:

| Engine | AIC |
|---|---|
| statsmodels | -2 lnL + 2k |
| Stata (`estat ic`) | -2 lnL + 2k, k = e(rank)+1 |
| R | -2 lnL + 2(k+1) - sigma^2 counts as a parameter |
| EViews | (-2 lnL + 2k) / n |

None is wrong. Comparing raw AIC across programs is meaningless unless you
know which convention each used - so EconEnv tells you.

```python
for note in comparison.explain():
    print('-', note)
```

## Standard errors, robust

```python
robust = econenv.compare_ols(df, 'y ~ x1 + x2', vcov='hc1')
robust.coefficients('std_err')
```
