# 5 - All four engines in one notebook

This is the MVP acceptance test from the project brief (section 46): a Python
cell, an R cell, a Stata cell and an EViews cell, executing in sequence, in
**one** Python kernel.

```python
%load_ext econenv
```

## 1 - Python

```python
import numpy as np
import pandas as pd
import econenv

rng = np.random.default_rng(20260904)
n = 120
df = pd.DataFrame({'x1': rng.normal(size=n), 'x2': rng.normal(size=n)})
df['y'] = 2.0 + 0.5 * df.x1 - 0.3 * df.x2 + rng.normal(scale=0.4, size=n)
df.head()
```

## 2 - R

```python
%%R -i df -o r_coef
fit <- lm(y ~ x1 + x2, data = df)
r_coef <- as.data.frame(coef(summary(fit)))
cat('R ok\n')
```

```python
r_coef
```

## 3 - Stata

```python
%stata_push df
```

```python
%%stata
regress y x1 x2
```

## 4 - EViews

```python
%%eviews -i df -q
equation eq1.ls y c x1 x2
```

```python
%eviews_get eq1.@r2
```

## 5 - back in Python, same objects

```python
print('df is unchanged:', df.shape)
%econ versions
```

Four engines. One kernel. One dataset. No CSV.
